**Stage 2 – Data Cleaning and Pre-processing**


**1. Handling Missing Values**

The dataset was checked for missing and null values using appropriate Python functions. The TotalCharges column was identified as a column that may contain blank values and was converted from object/string format to numeric format. Any resulting missing values were handled using an appropriate imputation method, such as replacing them with the median value.

This ensures that missing data does not affect statistical calculations and visualizations.

**2. Handling Duplicate Records**

The dataset was checked for duplicate rows using the duplicated() function. Duplicate records, if found, were removed to ensure that each customer was represented only once.

This prevents duplicate customers from affecting churn calculations and analysis.

**3. Column Name Cleaning and Unwanted Columns**

The column names were reviewed to ensure they were clear and consistent. Columns that were not useful for analytical purposes, such as the unique CustomerID, can be removed if they are not required for the analysis.

The remaining columns were retained because they provide useful information about customer demographics, services, contracts, billing, and churn behavior.

**4. Outlier Treatment**

Numerical columns such as Tenure, MonthlyCharges, and TotalCharges were examined using descriptive statistics and boxplots to identify potential outliers.

Extreme values were not automatically removed because they may represent genuine customer behavior. For example, a customer with high monthly or total charges may be a valid customer rather than an incorrect record.

Therefore, outliers were treated only when they were determined to be data errors or inappropriate values.

**5. Skewness Check and Transformation**

The numerical variables were checked for skewness to understand the distribution of the data.

Skewness was calculated using Python's skew() function. If a variable showed significant skewness, an appropriate transformation, such as a logarithmic transformation, could be applied.

For example, TotalCharges can be transformed using:

df['TotalCharges_log'] = np.log1p(df['TotalCharges'])

This can help reduce the effect of extreme values and improve the distribution of highly skewed variables.

**6. Data Type Conversion**

The data types of all columns were checked and converted where necessary.

In particular, TotalCharges was converted from a text/object data type into a numeric data type:
df['TotalCharges'] = pd.to_numeric(
    df['TotalCharges'],
    errors='coerce'
)

This allows numerical calculations, statistical analysis, and visualization to be performed correctly.

**7. Feature Transformations**

Additional features can be created from the existing variables to support deeper analysis.

For example, the Tenure variable can be divided into meaningful groups such as:

0–12 months – New customers
13–24 months – Short-term customers
25–48 months – Medium-term customers
49–72 months – Long-term customers

Similarly, MonthlyCharges can be categorized into Low, Medium, High, and Very High charge groups.

These derived features can help identify customer segments that have higher or lower churn rates.

**Final Data Quality Check**

After completing the cleaning and preprocessing steps, the dataset was checked again for:

Missing values
Duplicate records
Correct data types
Unwanted columns
Invalid or extreme values
Appropriate numerical distributions

The cleaned dataset is now structured and suitable for the next stage of the project, Exploratory Data Analysis (EDA).

The data cleaning and preprocessing stage ensures that the Telco Customer Churn dataset is clean, consistent, accurate, and analysis-ready. ### Chart 1: Customer Churn Distribution

```python
plt.figure(figsize=(7,5))
sns.countplot(data=df, x='Churn', color='steelblue')
plt.title('Customer Churn Distribution')
plt.xlabel('Churn Status')
plt.ylabel('Number of Customers')
plt.show()
```

**Interpretation:** The chart shows that customers who did not churn are much more numerous than customers who churned. This gives an overall view of the target variable and shows that churn is the smaller group in the dataset.

### Chart 2: Contract Type Distribution

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Contract', color='steelblue')
plt.title('Contract Type Distribution')
plt.xlabel('Contract Type')
plt.ylabel('Number of Customers')
plt.xticks(rotation=10)
plt.show()
```

**Interpretation:** Month-to-month contracts make up the largest group of customers, followed by one-year and two-year contracts. This distribution is useful when comparing churn across contract types.

### Chart 3: Customer Tenure Distribution

```python
plt.figure(figsize=(9,5))
sns.histplot(data=df, x='Tenure', bins=30, kde=True, color='steelblue')
plt.title('Distribution of Customer Tenure')
plt.xlabel('Tenure (Months)')
plt.ylabel('Number of Customers')
plt.show()
```

**Interpretation:** Customer tenure is spread across the full range from new customers to long-term customers. The distribution helps show how many customers are newer and how many have stayed for a longer period.

### Chart 4: Monthly Charges Distribution

```python
plt.figure(figsize=(9,5))
sns.histplot(data=df, x='MonthlyCharges', bins=30, kde=True, color='steelblue')
plt.title('Distribution of Monthly Charges')
plt.xlabel('Monthly Charges')
plt.ylabel('Number of Customers')
plt.show()
```

**Interpretation:** Monthly charges are spread across a wide range. Most customers fall within the middle charge range, while fewer customers have very low or very high monthly charges.

### Chart 5: Monthly Charges by Churn Status

```python
plt.figure(figsize=(7,5))
sns.boxplot(data=df, x='Churn', y='MonthlyCharges')
plt.title('Monthly Charges by Churn Status')
plt.xlabel('Churn Status')
plt.ylabel('Monthly Charges')
plt.show()
```

**Interpretation:** The box plot shows that customers who churned generally have higher monthly charges than customers who stayed. This indicates an association between higher monthly charges and churn in this dataset.

### Chart 6: Contract Type vs Customer Churn

```python
plt.figure(figsize=(9,5))
sns.countplot(data=df, x='Contract', hue='Churn')
plt.title('Contract Type vs Customer Churn')
plt.xlabel('Contract Type')
plt.ylabel('Number of Customers')
plt.legend(title='Churn')
plt.show()
```

**Interpretation:** Month-to-month customers show the largest number of churned customers. One-year and two-year contract customers show much lower churn counts relative to customers who stayed.

### Chart 7: Customer Tenure vs Churn

```python
plt.figure(figsize=(7,5))
sns.boxplot(data=df, x='Churn', y='Tenure')
plt.title('Customer Tenure by Churn Status')
plt.xlabel('Churn Status')
plt.ylabel('Tenure (Months)')
plt.show()
```

**Interpretation:** Customers who churned generally have shorter tenure than customers who stayed. This suggests that churn is associated with being a newer customer in this dataset.

### Chart 8: Internet Service vs Customer Churn

```python
df['ChurnNumeric'] = df['Churn'].map({'Yes': 1, 'No': 0})
internet_churn = df.groupby('InternetService')['ChurnNumeric'].mean().mul(100).reset_index()
plt.figure(figsize=(9,5))
sns.barplot(data=internet_churn, x='InternetService', y='ChurnNumeric', errorbar=None, color='steelblue')
plt.title('Churn Rate by Internet Service')
plt.xlabel('Internet Service')
plt.ylabel('Churn Rate (%)')
plt.show()
```

**Interpretation:** Fiber optic customers have the highest churn rate among the three internet-service groups, while customers without internet service have the lowest churn rate. These are associations in the dataset and do not by themselves show causation.

### Chart 9: Payment Method vs Customer Churn

```python
payment_churn = df.groupby('PaymentMethod')['ChurnNumeric'].mean().mul(100).reset_index()
plt.figure(figsize=(10,5))
sns.barplot(data=payment_churn, x='PaymentMethod', y='ChurnNumeric', errorbar=None, color='steelblue')
plt.title('Churn Rate by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Churn Rate (%)')
plt.xticks(rotation=20, ha='right')
plt.show()
```

**Interpretation:** Electronic check customers have the highest churn rate among the payment-method groups, while the automatic payment methods have lower churn rates. This shows a clear difference in churn rates across payment methods.

### Chart 10: Tenure vs Monthly Charges by Churn Status

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Tenure', y='MonthlyCharges', hue='Churn', alpha=0.6)
plt.title('Tenure vs Monthly Charges by Churn Status')
plt.xlabel('Tenure (Months)')
plt.ylabel('Monthly Charges')
plt.legend(title='Churn')
plt.show()
```

**Interpretation:** The scatter plot shows the relationship between tenure and monthly charges while separating customers by churn status. Churned customers appear more often among lower-tenure customers, while monthly charges vary across both churn groups.

### Chart 11: Contract Type & Tenure Group vs Churn

```python
df['TenureGroup'] = pd.cut(
    df['Tenure'],
    bins=[0, 12, 24, 48, 72],
    labels=['0-12 Months', '13-24 Months', '25-48 Months', '49-72 Months']
)
pivot = pd.pivot_table(df, values='ChurnNumeric', index='Contract', columns='TenureGroup', aggfunc='mean', observed=False) * 100
plt.figure(figsize=(10,6))
sns.heatmap(pivot, annot=True, fmt='.1f', cmap='YlOrRd')
plt.title('Churn Rate by Contract Type and Tenure Group')
plt.xlabel('Tenure Group')
plt.ylabel('Contract Type')
plt.show()
```

**Interpretation:** The heatmap shows how churn rates vary across both contract type and tenure group. Higher churn rates are concentrated in shorter-tenure and month-to-month combinations, while longer-term contract groups generally show lower churn rates.

### Chart 12: Contract Type & Internet Service vs Churn

```python
pivot_internet = pd.pivot_table(df, values='ChurnNumeric', index='Contract', columns='InternetService', aggfunc='mean') * 100
plt.figure(figsize=(10,6))
sns.heatmap(pivot_internet, annot=True, fmt='.1f', cmap='Blues')
plt.title('Churn Rate by Contract Type and Internet Service')
plt.xlabel('Internet Service')
plt.ylabel('Contract Type')
plt.show()
```

**Interpretation:** The heatmap compares churn rates across two categorical variables at the same time. Month-to-month contracts generally show higher churn rates across internet-service categories, while longer contracts show lower rates.

### Chart 13: Churn Rate by Contract Type

```python
contract_churn = df.groupby('Contract')['ChurnNumeric'].mean().mul(100).reset_index()
plt.figure(figsize=(8,5))
sns.barplot(data=contract_churn, x='Contract', y='ChurnNumeric', errorbar=None, color='steelblue')
plt.title('Churn Rate by Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Churn Rate (%)')
plt.show()
```

**Interpretation:** Month-to-month contracts have the highest churn rate, followed by one-year contracts, while two-year contracts have the lowest churn rate. The chart clearly shows a strong association between contract type and churn.

### Chart 14: Correlation Heatmap

```python
# Ensure TotalCharges is numeric and handle missing values
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'] = df['TotalCharges'].fillna(0)

numeric_cols = ['Tenure', 'MonthlyCharges', 'TotalCharges', 'ServiceCount', 'ChurnNumeric']
correlation = df[numeric_cols].corr()
plt.figure(figsize=(8,6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Correlation Heatmap')
plt.show()
```

**Interpretation:** Tenure has a negative correlation with churn, while MonthlyCharges has a positive correlation with churn. The correlations indicate association between these numerical variables and churn, but correlation does not establish causation.

2. **Strengthen early customer onboarding:** Give additional attention to new customers during the first several months because churn is more common among lower-tenure customers.
3. **Review higher-charge customer segments:** Investigate whether customers with higher monthly charges receive sufficient value from their services and consider suitable retention offers.
4. **Investigate fiber optic customer experience:** Review pricing, service quality, support experience, and customer satisfaction for fiber optic customers because this group shows relatively high churn.
5. **Encourage convenient automatic payment methods:** Investigate the reasons for high churn among electronic-check customers and promote suitable automatic payment options where appropriate.
6. **Develop a churn early-warning process:** Use the dashboard filters and customer-level data to identify high-risk segments for proactive retention activity.

These recommendations are based on observed relationships in the dataset and should be validated with additional business and customer-experience information before implementation.
